```python
import pandas as pd
import numpy as np
```

```python
#Sample info based on Table S2 https://www.biorxiv.org/content/biorxiv/early/2022/05/24/2022.05.24.493188/DC3/embed/media-3.xlsx?download=true
sample_table = pd.read_table("samples.txt", header=None)
sample_table.columns=["colname", "si"]
sample_table["Rep"] = sample_table["colname"].str.split("_").str[-1]
sample_table["id"] = sample_table["colname"].str.split("_").str[-2]
sample_table["Sample"] = sample_table["si"] + "_" + sample_table["colname"]
sample_table.to_csv("sample_all.txt", sep="\t", index=False)
sample_table.head()
```

```python
#Clusters from Table S4 https://www.biorxiv.org/content/biorxiv/early/2022/05/24/2022.05.24.493188/DC5/embed/media-5.xlsx?download=true
clusters = pd.read_excel("media-5.xlsx")
clusters = clusters.rename(columns={"Unnamed: 0": "Gene", "NIPBL7": "NIPBL", 
                                    "Rad21": "RAD21", "Rad21_NIPBL": "RAD21.NIPBL", 
                                    "Mau2": "MAU2"})
clusters.head()
```

```python
#Counts from GEO https://ftp.ncbi.nlm.nih.gov/geo/series/GSE196nnn/GSE196449/suppl/GSE196449%5FAllsamples.GRCh38.xlsx
counts = pd.read_excel("GSE196449_Allsamples.GRCh38.xlsx")
counts = counts.rename(columns={"Unnamed: 0": "Gene", "Unnamed: 1": "ensembl_id"})
counts = counts.rename(columns=dict(zip(sample_table.colname,sample_table.Sample)))
counts.head()
```

```python
counts = pd.merge(counts, clusters, on="Gene", how="left")
counts["Cluster"] = np.where(np.isnan(counts["Cluster"]), "None", counts["Cluster"])
counts.head()
```

```python
counts.to_csv("GSE196449_Allsamples.GRCh38_counts_processed_all.txt", sep="\t", index=False)
```
